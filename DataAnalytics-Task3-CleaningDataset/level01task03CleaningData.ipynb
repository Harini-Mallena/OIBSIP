## 3.Data Handling
Missing values can affect statistical analysis and machine learning models. Therefore, the missing values identified during the data quality assessment are handled using appropriate techniques based on the data type and the proportion of missing values.


### Missing Value Analysis
The number and percentage of missing values are calculated for each column to determine the extent of missing data before appplying any cleaning technique.

```python
missing_report = pd.DataFrame({
    "Missing_Count":df.isnull().sum(),
    "Missing_Percentage":(df.isnull().mean()*100).round(2)
})
missing_report
```

#### Missing value treatment decisions
The missing values are handled according to the data type and the proportion of missing values in each column.
- Age: Missing numerical values are replaced with the median because the median is less affected by extreme values.
- Embarked: Missing categorical values are replaced with the mode, which represents the most frequently occuring category.
- Cabin: The column contains a large proportion of missing values.
  
Since reliable imputation is difficult, the column is removed.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop(columns=["Cabin"])
```

#### Verification after missing value treatment
After applying the selected treatment methods, the dataset is checked again to confirm that the missing values have been handled successfully.

```python
missing_after = pd.DataFrame({
    "Missing_Count":df.isnull().sum(),
    "Missing_percentage":(df.isnull().mean()*100).round(2)
})
missing_after
```

## 6.Outlier Detection
Outliers are observations that differ substantially from the majority of values in a numerical column. The Interquartile Range (IQR) method is used to identify potential outliers.

IQR = Q3-Q1

Lower Bound = Q1-1.5*IQR

Upper Bound = Q3+1.5*IQR

Values below the lower bound or above the upper bound are considered potential outliers.

### Identifying Potential Outliers
The IQR  method is applied to the numerical columns in the dataset.
The number of potential outliers is calculated for each numerical column.

```python
numeric_columns = df.select_dtypes(include=np.number).columns
outlier_report = []
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3-Q1
    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR
    outliers = df[
    (df[column]<lower_bound)|(df[column]>upper_bound)
    ]
    outlier_report.append({
    "Column":column,
    "Q1":Q1,
    "Q3":Q3,
    "IQR":IQR,
    "Lower_Bound":lower_bound,
    "Upper_Bound":upper_bound,
    "Outlier_Count":len(outliers)
    })
outlier_report = pd.DataFrame(outlier_report)
outlier_report
```

### Outlier Treatment Decision

The IQR method identified potential outliers in Age, SibSp, Parch and
Fare. However, statistical outliers are not automatically considered
data errors.

The identified observations were reviewed based on the meaning of each
variable. The extreme values can represent genuine passenger
characteristics, such as older passengers, larger family groups, or
higher ticket fares.

Therefore, the identified outliers are retained rather than removed or
capped. This prevents the loss of valid information from the dataset.

### Outlier Summary

No observations are removed solely because they were identified as
statistical outliers using the IQR method. The outlier detection results
are documented for transparency, while valid extreme observations are
retained for further analysis.

## 7.Data Type Correction

The data types of all columns are reviewed to ensure that they are
appropriate for their intended use.

Identifier columns are treated as strings, numerical measurements are
stored using appropriate numeric types, and categorical columns are
stored as text values.

### Current Data Types

The current data types are reviewed before applying any necessary
corrections.

```python
df.dtypes
```

### PassengerId Data Type

PassengerId is an identifier and does not represent a numerical
quantity. Therefore, it is converted from an integer to a string so
that it is not accidentally treated as a numerical measure during
analysis.

```python
df["PassengerId"] = df["PassengerId"].astype(str)
```

### Numerical Column Correction

The numerical columns are explicitly converted to appropriate numeric
data types to ensure that mathematical and statistical operations can
be performed correctly.

```python
numeric_columns = ["Survived","Pclass","Age","SibSp","Parch","Fare"]
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

### Categorical Column Correction

Categorical columns are stored as text values so that categories such
as gender and embarkation port are handled consistently.

```python
categorical_columns = ["Name", "Sex", "Ticket", "Embarked"]
for column in categorical_columns:
    df[column] = df[column].astype(str)
```

### Verification of Data Types

The data types are checked again after correction to confirm that the
columns have appropriate data types.

## 8.Before vs After Cleaning Summary

```python
print("Current number of rows:",len(df))
```

### Cleaning Summary Table

```python
expected_types = {
    "PassengerId": "string",
    "Survived": "numeric",
    "Pclass": "numeric",
    "Name": "string",
    "Sex": "string",
    "Age": "numeric",
    "SibSp": "numeric",
    "Parch": "numeric",
    "Ticket": "string",
    "Fare": "numeric",
    "Embarked": "string"
}
correct_types = 0
for column, expected in expected_types.items():
    actual = str(df[column].dtype)
    if expected == "string" and actual in ["object","string","str"]:
        correct_types += 1
    elif expected == "numeric" and pd.api.types.is_numeric_dtype(df[column]):
        correct_types += 1
total_columns = len(expected_types)
summary_table = pd.DataFrame({
    "Metric": [
        "Total Null Values",
        "Duplicate Rows",
        "Number of Rows",
        "Number of Columns",
        "Correct Data Types"
    ],
    "After Cleaning": [
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df.shape[0],
        df.shape[1],
        f"{correct_types}/{total_columns}"
    ]
})
summary_table
```
